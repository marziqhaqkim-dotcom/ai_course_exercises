```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Define the weather-related keywords in Swedish
weather_keywords_swe = ['snö', 'is', 'regn', 'storm', 'vind', 'väder']

# Prepare a dictionary to store monthly counts for all keywords
monthly_weather_alerts = {}

# Add 'snö' alerts (already calculated in sno_alerts_per_month) to the dictionary
monthly_weather_alerts['snö'] = sno_alerts_per_month

# Calculate monthly alerts for other weather-related keywords
for keyword in weather_keywords_swe:
    if keyword == 'snö':
        continue  # Skip 'snö' as it's already handled

    df_keyword_filtered = df_2024[
        df_2024['description_text2'].str.contains(keyword, case=False, na=False) |
        df_2024['header_text2'].str.contains(keyword, case=False, na=False)
    ].copy()

    if not df_keyword_filtered.empty:
        df_keyword_filtered['start_month'] = df_keyword_filtered['start'].dt.to_period('M')
        keyword_alerts_per_month = df_keyword_filtered['start_month'].value_counts().sort_index()
        keyword_alerts_per_month.index = keyword_alerts_per_month.index.to_timestamp()
        monthly_weather_alerts[keyword] = keyword_alerts_per_month
    else:
        # If no alerts for a keyword, create an empty Series with same index as others for consistent plotting
        if not monthly_weather_alerts['snö'].empty:
            empty_series = pd.Series(0, index=monthly_weather_alerts['snö'].index)
            monthly_weather_alerts[keyword] = empty_series
        else:
            # Handle case where 'snö' alerts also empty (unlikely but for robustness)
            monthly_weather_alerts[keyword] = pd.Series()

# Convert the dictionary to a DataFrame
df_monthly_weather_alerts = pd.DataFrame(monthly_weather_alerts)

# Plotting all weather-related keywords
plt.figure(figsize=(14, 7))
sns.lineplot(data=df_monthly_weather_alerts, markers=True)
plt.title('Monthly Alerts for Various Weather-Related Keywords in 2024')
plt.xlabel('Month')
plt.ylabel('Number of Alerts')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.legend(title='Keyword')
plt.tight_layout()
plt.show()
```
