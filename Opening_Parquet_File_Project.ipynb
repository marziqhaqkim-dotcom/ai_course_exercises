```python
import pandas as pd

# Replace 'your_file.parquet' with the actual name of your file
file_name = 'service_alerts_2024 (1).parquet' # Corrected file name

df = pd.read_parquet(file_name)

# View the first few rows
df.head()
```

```python
display(df.describe())
```

```python
missing_values = df.isnull().sum()
display(missing_values)
```

```python
df = df.drop(columns=['alert_message_timestamp'])
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='cause', hue='cause', palette='viridis', legend=False)
plt.title('Distribution of Cause Column')
plt.xlabel('Cause')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Convert 'start' and 'end' columns to datetime objects if they aren't already
df['start'] = pd.to_datetime(df['start'])
df['end'] = pd.to_datetime(df['end'])

# Find the minimum 'start' date and maximum 'end' date
min_start_date = df['start'].min()
max_end_date = df['end'].max()

print(f"Earliest start date: {min_start_date}")
print(f"Latest end date: {max_end_date}")
```

```python
# Calculate the duration of the data
time_delta = max_end_date - min_start_date
print(f"Total time span of the data: {time_delta}")

# Calculate the number of months covered
# This is an approximation, as it counts month boundaries
num_months = (max_end_date.year - min_start_date.year) * 12 + (max_end_date.month - min_start_date.month) + 1

print(f"The data spans approximately {num_months} months.")
```

```python
# Extract the year from the 'end' column and count occurrences
end_year_counts = df['end'].dt.year.value_counts().sort_index()
display(end_year_counts)
```

```python
# Filter the DataFrame to keep only rows where the 'end' year is 2024
df_2024 = df[df['end'].dt.year == 2024]

print(f"Original DataFrame shape: {df.shape}")
print(f"Filtered DataFrame shape (only 2024 data): {df_2024.shape}")
display(df_2024.head())
```

```python
# Extract the month from the 'start' column
df_2024 = df_2024.copy()
df_2024['start_month'] = df_2024['start'].dt.to_period('M')

# Count the number of alerts per month
alerts_per_month = df_2024['start_month'].value_counts().sort_index()

# Display the results
display(alerts_per_month)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Convert the PeriodIndex to datetime for better plotting
alerts_per_month.index = alerts_per_month.index.to_timestamp()

plt.figure(figsize=(12, 6))
sns.lineplot(x=alerts_per_month.index, y=alerts_per_month.values)
plt.title('Monthly Alert Counts (2023-2024)')
plt.xlabel('Month')
plt.ylabel('Number of Alerts')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.countplot(data=df_2024, x='cause', hue='cause', palette='viridis', legend=False)
plt.title('Distribution of Cause Column for 2024 Data')
plt.xlabel('Cause')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Group by 'cause' and 'stop_id', then count occurrences
# Unstack 'stop_id' to make it easier to get top N per cause
# Fill NaN with 0 for causes that don't have certain stop_ids
# Apply nlargest to get top 5 stop_ids for each cause
# Stack it back to a Series for a cleaner display
top_stop_ids_per_cause = df_2024.groupby(['cause', 'stop_id']).size().unstack(fill_value=0)

# Get the top 5 stop_ids for each cause
result = top_stop_ids_per_cause.apply(lambda x: x.nlargest(5).index.tolist(), axis=1)

print("Top 5 'stop_id' locations per 'cause':")
display(result)
```
